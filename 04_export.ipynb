### Pandas Export
#### to csv, to excel, to html, to json, to sql


```python
df=pd.DataFrame()
```

# to_csv

```python
df = pd.read_csv('deliveries.csv')
df.head()
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df
```

```python
# temp_df.to_csv('batsman_runs.csv')
temp_df.to_csv('batsman_runs.csv', index=False)
```

```python
pd.pivot_table(df, index='batsman', columns='bowling_team', values='batsman_runs', aggfunc='sum').to_csv('batsman_vs_team.csv')
```

# to_excel

```python
df
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df
```

```python
temp_df.to_excel('batsman_runsss.xlsx')
```

```python
temp_df.to_excel('batsman_runsss.xlsx', sheet_name='batsman_run_ipl')
```

```python
temp_df2 = pd.pivot_table(df, index='batsman', columns='bowling_team', values='batsman_runs', aggfunc='sum')
```

```python
with pd.ExcelWriter('output.xlsx') as writer:
    temp_df.to_excel(writer, sheet_name='Sheet_name_1')
    temp_df2.to_excel(writer, sheet_name='Sheet_name_2')
```

# to_html
#### dataframe ko html tables m convert
#### aap apne analysis ko blog k form m analyse krna chahte ho or blog m aapko annalysis to as a table print krna chahte ho

```python
df
```

```python
df.query('batsman_runs == 6').pivot_table(index='over', columns='ball', values='batsman_runs', aggfunc='count').to_html('sixes_heatmap.html')
```

# to_json

```python
df
```

```python
df.groupby(['batting_team', 'batsman'], as_index=False)['batsman_runs'].sum()    # as_index=False se DataFrame bn jaayega ye series
```

```python
df.groupby(['batting_team', 'batsman'])['batsman_runs'].sum().unstack()    # unstack() default me last level (batsman) ko columns bana deta hai.
```

```python
df.groupby(['batting_team', 'batsman'])['batsman_runs'].sum().unstack().to_json('ipl.json')
```
